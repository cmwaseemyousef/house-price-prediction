```python
# 01_data_cleaning.ipynb
import pandas as pd

# Load the dataset
data = pd.read_csv("../data/house_prices.csv")

# Perform basic data cleaning
# Drop rows with missing values
data = data.dropna()

# Drop columns that are not needed
data = data.drop(columns=['Column1', 'Column2'])  # Replace 'Column1', 'Column2' with actual columns to drop

# Save the cleaned data for later use
data.to_csv("../data/cleaned_house_prices.csv", index=False)

# Show the first few rows of the cleaned data
data.head()
```
